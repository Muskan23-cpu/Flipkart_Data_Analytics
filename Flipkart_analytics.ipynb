```python
# sales by category
sales=df.groupby("category")["units_sold"].sum().sort_values(ascending=False)
sales.plot(kind="bar",figsize=(10,5))
plt.title("Units sold by category")
plt.xlabel("Category")
plt.ylabel("Units Sold")
plt.show()
```

```python
# avg final price by category
avg=df.groupby("category")["final_price"].mean().sort_values(ascending=False)
avg.plot(kind="bar",figsize=(10,5),color=["skyblue"])
plt.title("Avg price by category")
plt.xlabel("Category")
plt.ylabel("Avg Price")
plt.show()
```

```python
#Top 10 brands by units sold
top_brands=df.groupby("brand")["units_sold"].sum().sort_values(ascending=False).head(10)
top_brands.plot(kind="bar",figsize=(10,5),color=["green"])
plt.title("Top 10 Brands by Units Sold")
plt.xlabel("Brand")
plt.ylabel("Units Sold")
plt.show()
```

```python
# Delivery days distribution
df['delivery_days'].plot(kind="hist",edgecolor="white",figsize=(10,5))
plt.title("Delivery Days Distribution")
plt.xlabel("Delivery Days")
plt.ylabel("Number of Products")
plt.show()
```

```python
# Payment mode distribution
payment=df["payment_modes"].value_counts()
payment.plot(kind="bar",figsize=(8,5),color="brown")
plt.title("Payment Mode Distribution")
plt.xlabel("Payment modes")
plt.ylabel("Number of Products")
plt.show()
```

```python
# Product Category Distribution
df["category"].value_counts().plot(kind="pie",autopct='%1.1f%%',figsize=(7,5))
plt.title("Product Distribution by Category")
plt.show()
```

```python
#Returnable vs Non-eturnable
df['is_returnable'].value_counts().plot(
    kind='pie',
    autopct='%1.1f%%',
    figsize=(7,4)
)

plt.title('Returnable vs Non-Returnable Products')
plt.show()
```
